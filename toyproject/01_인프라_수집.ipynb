## 수집 실행

범위가 좁아서(정문 반경 1km) 몇 분 안에 끝날 것으로 예상한다. 중간에 멈춰도 다시 실행하면 끝난 영역은 건너뛴다.

**카테고리 선정에 대한 주의**

지금 아래에서 수집하는 카테고리(병원·약국·편의점·대형마트·은행·지하철역·음식점·카페 8개)는 카카오 로컬 API가 제공하는 18개 카테고리 중 임의로 좁힌 목록.      학교, 어린이집·유치원, 문화시설, 공공기관처럼 실제 보행 목적지인데 빠진 카테고리가 있고, "보행자가 실제 목적지로 걸어가는 장소인가"라는 기준으로 항목 하나하나를 따져서 고른 것도 아니다.

지금은 수집→저장→확인까지 파이프라인 전체가 정상 동작하는지 보는 예비(pilot) 실행이라 이 8개 카테고리 그대로 진행한다. 본 수집 단계에서는 카테고리 전체(18개)를 받거나, 최소한 위에서 빠진 목적지형 카테고리를 추가해서 다시 실행해야 한다.

## 결과 불러오기 및 확인

```python
records = [json.loads(line) for line in open(infra.OUT_PATH, encoding="utf-8")]
df = pd.DataFrame(records)
df = df.drop_duplicates(subset="id")  # 인접 격자 경계에 걸쳐 중복 수집된 장소 제거

df["x"] = df["x"].astype(float)  # 경도
df["y"] = df["y"].astype(float)  # 위도

print(f"총 {len(df):,}건")
df["category_group_name"].value_counts()
```

```python
df[["place_name", "category_group_name", "road_address_name", "x", "y"]].head(10)
```

## 지도로 확인



```python
try:
    import folium
    from IPython.display import display

    m = folium.Map(location=[infra.CENTER_LAT, infra.CENTER_LON], zoom_start=15)
    folium.Marker(
        [infra.CENTER_LAT, infra.CENTER_LON],
        popup="중앙대 정문",
        icon=folium.Icon(color="red"),
    ).add_to(m)

    for _, row in df.iterrows():
        folium.CircleMarker(
            [row["y"], row["x"]],
            radius=3,
            popup=f"{row['place_name']} ({row['category_group_name']})",
        ).add_to(m)

    display(m)
except ImportError:
    print("folium 미설치 -- pip install folium 후 다시 실행")
```
